```python
#create a function to change temperature from celsius to fahrenheit

"""
fahrenheit = (celsius * 9/5) + 32


(32°F − 32) × 5/9 = 0°C
"""

# name = 'Joseph' #global variable

def celsius_to_fahrenheit(celsius):
  fahrenheit = (celsius * 9/5) + 32

  return (fahrenheit, celsius)

print(f'{celsius_to_fahrenheit(32)[1]} degrees celsius is {celsius_to_fahrenheit(32)[0]} Fahrenheit')
```

```python
def celsius_to_fahrenheit_1(celsius = 32):

  fahrenheit = (celsius * 9/5) + 32

  return (fahrenheit, celsius)

```

```python
test = celsius_to_fahrenheit_1()

print(f'{test[1]} degrees celsius is {test[0]} Fahrenheit')
```

### Lambda functions
They are small, anonymous, one line functions.


```python
numbers = [2, 4, 6, 8, 10]


def square(x):
  print (x ** 2)


for number in numbers:
  square(number)
```

```python
numbers = [2, 4, 6, 8, 10]



test_lambda = lambda number: number ** 2

test_lambda(2)
```

```python
#normal named function

def square(x):
  print (x ** 2)


#its lambda equivalent
lambda number: number ** 2

```
