```python
#while loop
count = 1

while count <= 5:
    print(count)
    count += 1

# count <= 5

```

```python
#while loop
count = 1

while count <= 5:
  print(f"Current value of count is: {count}")
  count += 1
  print(f"Updated value of count is {count}")

```
